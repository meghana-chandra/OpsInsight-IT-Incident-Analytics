```python
# Prepare features and target

X = incident_level.drop(
    columns=["number", "opened_at", "made_sla"]
)

y = incident_level["made_sla"].astype(int)

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = [
    "contact_type",
    "location",
    "category",
    "subcategory",
    "impact",
    "urgency",
    "priority",
    "assignment_group",
    "opened_by",
    "notify"
]

numerical_features = [
    "knowledge",
    "u_priority_confirmation",
    "opened_hour",
    "opened_day",
    "opened_month",
    "opened_weekday",
    "is_weekend"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)

print("Preprocessor created successfully.")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

# Combine preprocessing and model
rf_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", rf_model)
    ]
)

# Train the model
rf_pipeline.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
# Predict SLA outcome for the unseen test data

y_pred = rf_pipeline.predict(X_test)

print("First 20 predictions:")
print(y_pred[:20])

print("\nActual values:")
print(y_test.iloc[:20].values)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Model Performance")
print("-----------------")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# aggregate feature importance to original columns

original_features = []

for feature in feature_importance["feature"]:
    feature = feature.replace("categorical__", "")
    feature = feature.replace("numerical__", "")
    
    # Match against our actual original feature names
    matched = None
    
    for col in X.columns:
        if feature == col or feature.startswith(col + "_"):
            matched = col
            break
    
    original_features.append(matched)

feature_importance["original_feature"] = original_features

aggregated_importance = (
    feature_importance
    .groupby("original_feature")["importance"]
    .sum()
    .sort_values(ascending=False)
)

print(" Aggregated Feature Importance")
print("----------------------------------------")
display(aggregated_importance.to_frame("importance"))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Categorical features
categorical_features = [
    "contact_type",
    "location",
    "category",
    "subcategory",
    "impact",
    "urgency",
    "priority",
    "assignment_group",
    "opened_by",
    "notify"
]

# Numerical features
numerical_features = [
    "knowledge",
    "u_priority_confirmation",
    "opened_hour",
    "opened_day",
    "opened_month",
    "opened_weekday",
    "is_weekend"
]

# Preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)

print("Preprocessor created successfully.")
print("\nCategorical features:", len(categorical_features))
print("Numerical features:", len(numerical_features))
```

```python
import numpy as np
from sklearn.metrics import precision_score, recall_score, f1_score

# Probability that each incident will breach SLA
breach_prob = rf_pipeline_corrected.predict_proba(X_test)[:, 0]

threshold_results = []

for threshold in np.arange(0.20, 0.61, 0.05):
    
    predicted_breach = (breach_prob >= threshold).astype(int)
    
    actual_breach = 1 - y_test
    
    precision = precision_score(
        actual_breach,
        predicted_breach,
        zero_division=0
    )
    
    recall = recall_score(
        actual_breach,
        predicted_breach,
        zero_division=0
    )
    
    f1 = f1_score(
        actual_breach,
        predicted_breach,
        zero_division=0
    )
    
    threshold_results.append({
        "threshold": round(threshold, 2),
        "precision": round(precision, 4),
        "recall": round(recall, 4),
        "f1": round(f1, 4)
    })

threshold_df = pd.DataFrame(threshold_results)

display(threshold_df)
```

```python
query = """
WITH category_summary AS (
    SELECT
        category,
        COUNT(*) AS total_incidents,
        SUM(CASE WHEN made_sla = 0 THEN 1 ELSE 0 END) AS sla_breached
    FROM incidents
    GROUP BY category
)

SELECT
    category,
    total_incidents,
    sla_breached,
    ROUND(
        100.0 * sla_breached / total_incidents,
        2
    ) AS breach_rate_pct
FROM category_summary
WHERE total_incidents >= 100
ORDER BY breach_rate_pct DESC;
"""

result = pd.read_sql_query(query, conn)

display(result)
```

```python
query = """
WITH category_summary AS (
    SELECT
        category,
        COUNT(*) AS total_incidents,
        SUM(CASE WHEN made_sla = 0 THEN 1 ELSE 0 END) AS sla_breached
    FROM incidents
    GROUP BY category
)

SELECT
    category,
    total_incidents,
    sla_breached,
    ROUND(
        100.0 * sla_breached / total_incidents,
        2
    ) AS breach_rate_pct,
    RANK() OVER (
        ORDER BY
            100.0 * sla_breached / total_incidents DESC
    ) AS breach_rank
FROM category_summary
WHERE total_incidents >= 100
ORDER BY breach_rank;
"""

result = pd.read_sql_query(query, conn)

display(result)
```
